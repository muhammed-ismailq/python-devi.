```python
# ── KısIM 1: Veri Yapıları ve Temel Mantık ─────────────────

# Değişken Tanımlama
ad_soyad = "Ayşe Yılmaz"        # String (metin) tipi
aylik_ucret = 650.0            # Float (ondalıklı sayı) tipi
sadakat_ayi = 30               # Integer (tam sayı) tipi
aktif_mi = True                # Boolean (mantıksal) tipi

# Veri tiplerini yazdırma
print(f"Ad Soyad   : {ad_soyad}  → tip: {type(ad_soyad).__name__}")
print(f"Aylık Ücret: {aylik_ucret} → tip: {type(aylik_ucret).__name__}")
print(f"Sadakat Ay : {sadakat_ayi}    → tip: {type(sadakat_ayi).__name__}")
print(f"Aktif Mi   : {aktif_mi}  → tip: {type(aktif_mi).__name__}")
```

```python
# Liste Kullanımı: Şirketin sunduğu 5 hizmet
hizmetler = [
    "Fiber Internet",
    "Mobil Hat",
    "IPTV",
    "Bulut Yedekleme",
    "Siber Güvenlik Paketi"
]

# Sözlük Kullanımı: Müşterinin tüm bilgileri key-value formatında
musteri = {
    "ad_soyad"    : ad_soyad,
    "aylik_ucret" : aylik_ucret,
    "sadakat_ayi" : sadakat_ayi,
    "aktif_mi"    : aktif_mi,
    "hizmetler"   : hizmetler
}

print("Hizmet Listesi:", hizmetler)
print("Müşteri Sözlüğü:", musteri)
```

```python
# Koşullu İfade (If-Else): VIP müşteri kontrolü
if musteri["aylik_ucret"] > 500 or musteri["sadakat_ayi"] > 24:
    print("VIP Müşteri: İndirim Tanımla")   # OR operatörü: iki koşuldan biri yeterlİ
else:
    print("Standart Müşteri")
```

```python
import random                    # random kütüphanesi: rastgele sayı üretimi

# String Operasyonu: adı büyük harfe çevirme
ad_buyuk = musteri["ad_soyad"].upper()
print(f"Büyük Harf Ad: {ad_buyuk}")

# customerID formatında ID oluşturma: IST-2026-XXXXX
rastgele_sayi = random.randint(10000, 99999)   # 5 haneli rastgele sayı
customer_id = f"IST-2026-{rastgele_sayi}"          # f-string ile birleştirme
print(f"Müşteri ID   : {customer_id}")
```

❓ Kritik Soru: Neden Liste yerine Sözlük?

Liste, verileri sıra numarası (index) ile saklar: musteri[0], musteri[2]… Hangi indeksin neyi temsil ettiğini hatırlamak zorunda kalırsınız ve kodun okunması güçleşir.

Sözlük ise verileri anlamlı bir anahtar (key) ile saklar: musteri["ad_soyad"], musteri["aylik_ucret"]. Avantajlar: (1) Kod kendini açıklar — musteri["sadakat_ayi"] anında ne olduğunu söyler. (2) Yeni alan eklemek kolaylaşır, sıra bozulmaz. (3) Arama O(1) zamanlı ve anlamlıdır. Gerçek dünyada bir müşteri "kaydı" mantıken sözlük yapısına uygundur.

```python
# ── KısIM 2: Fonksiyonlar, Döngüler ve Kütüphaneler ────────

import math                      # math kütüphanesi: matematiksel işlemler
import random                    # random kütüphanesi: rastgele sayı üretimi
from datetime import date         # datetime kütüphanesi: tarih işlemleri

# Ana Liste: 5 müşterinin bilgileri sözlük olarak tutulur
musteriler_listesi = [
    {"ad": "Ahmet Kaya",    "aylik_ucret": 450.0, "sadakat_ayi": 6,  "aktif_mi": True},
    {"ad": "Fatma Demir",  "aylik_ucret": 720.0, "sadakat_ayi": 36, "aktif_mi": True},
    {"ad": "Mehmet Çelik", "aylik_ucret": 300.0, "sadakat_ayi": 3,  "aktif_mi": False},
    {"ad": "Zeynep Arslan","aylik_ucret": 890.0, "sadakat_ayi": 48, "aktif_mi": True},
    {"ad": "Can Şahin",    "aylik_ucret": 150.0, "sadakat_ayi": 2,  "aktif_mi": False},
]
```

```python
# Fonksiyon Tanımlama: aylık ücrete %20 KDV ekler
def tutar_hesapla(aylik_ucret):
    kdv_orani = 1.20                          # %20 KDV çarpanı
    kdvli_tutar = aylik_ucret * kdv_orani    # Çarpma işlemi (Aritmetik Operatör)
    return kdvli_tutar                        # Fonksiyon geri dönüş değeri
```

```python
bugun = date.today()                         # datetime kütüphanesi: bugünün tarihi
toplam_fatura = 0

print("\n===== MÜŞTERİ FATURA RAPORU =====")
print(f"Tarih: {bugun}\n")

# For Döngüsü: her müşteri için fatura hesapla ve Churn riski tespit et
for musteri in musteriler_listesi:         # Döngü Kullanımı
    kdvli = tutar_hesapla(musteri["aylik_ucret"])  # Fonksiyon Çağırma
    yuvarlanmis = math.ceil(kdvli)           # math kütüphanesi: yukarı yuvarlama
    toplam_fatura += yuvarlanmis             # Kümülatif toplam (Birikimli Atama)

    # Churn (ayrılma) riski tespiti: Koşullu İfade
    # Kural: aktif değil VEYA (düşük ücret VE az sadakat) → Churn riski var
    if not musteri["aktif_mi"] or \
       (musteri["aylik_ucret"] < 400 and musteri["sadakat_ayi"] < 12):
        churn_durum = "⚠ CHURN RİSKİ"
    else:
        churn_durum = "✓ Güvenli"

    print(f"{musteri['ad']:<20} | KDVli: {yuvarlanmis:>7} TL | {churn_durum}")

print(f"\nToplam Fatura Tutarı : {toplam_fatura} TL")
print(f"Fatura Tarihi        : {bugun}")     # datetime kütüphanesi: tarih yazdırma
```

```python
# Set Kullanımı: tekrarlı hizmet listesinden benzersizleri çıkarma
tum_hizmetler = [                            # Liste Tanımlama (tekrarlı)
    "Fiber Internet", "Mobil Hat", "IPTV",
    "Fiber Internet", "Bulut Yedekleme",     # Tekrarlı eleman
    "Mobil Hat", "Siber Güvenlik Paketi",    # Tekrarlı eleman
    "IPTV", "Bulut Yedekleme"               # Tekrarlı eleman
]

benzersiz_hizmetler = set(tum_hizmetler)  # set() ile tekrarları temizle
print("\nBenzersiz Hizmetler:", benzersiz_hizmetler)
print(f"Toplam hizmet sayısı: {len(benzersiz_hizmetler)}")
```

❓ Kritik Soru: Churn Riski Nasıl Tespit Edildi?

For döngüsü her müşteri sözlüğünü sırayla gezerken üç göstergeye bakılır:

1. Aktiflik durumu: aktif_mi == False olan müşteri zaten hizmeti kullanmıyor — en güçlü churn sinyali.
2. Düşük ücret + kısa sadakat: aylik_ucret < 400 TL ve sadakat_ayi < 12 birlikte gerçekleşiyorsa müşteri henüz bağ kurmamış ve az ödüyor — ayrılması kolay.
3. OR mantığı: İki koşuldan biri yeterliydi — birincisi tek başına churn, ikincisi istatistiksel "risk profili".

Örnek: Mehmet Çelik (aktif değil) ve Can Şahin (150 TL, 2 ay) bu döngüde "⚠ CHURN RİSKİ" olarak işaretlendi.
